```python
def solve(M, items):
    n=len(items)
    OFFSET=n

    dp=[[-1]*(2*n+1) for _ in range(M+1)]
    choice=[[None]*(2*n+1) for _ in range(M+1)]
    dp[0][OFFSET]=0

    for i, (p, h, t) in enumerate(items):
        new_dp = [row[:] for row in dp]

        for j in range(M+1):
            for d in range(2*n+1):
                if dp[j][d]==-1:
                    continue
                if j+p>M:
                    continue

                if t=='A':
                    nd=d-1
                else:
                    nd=d+1

                if 0<=nd<2*n+1:
                    val=dp[j][d]+h
                    if val>new_dp[j+p][nd]:
                        new_dp[j+p][nd]=val
                        choice[j+p][nd]=(j, d, i)
        dp=new_dp

    best_val=0
    best_state=None

    for j in range(M+1):
        for d in range(OFFSET, 2 * n + 1):
            if dp[j][d]>best_val:
                best_val=dp[j][d]
                best_state=(j,d)

    if best_val==0:
        print(0)
        return

    # 回溯
    res=[]
    j,d=best_state

    while choice[j][d] is not None:
        pj, pd, idx = choice[j][d]
        res.append(idx)
        j,d=pj,pd

    res.reverse()
    res = [x + 1 for x in res]


    print("最大快乐值:", best_val)
    print("选择商品编号:", res)

items = [
    (3, 5, 'A'),
    (4, 7, 'B'),
    (2, 4, 'A'),
    (5, 8, 'B'),
    (3, 6, 'B'),
]

print("M = 10")
solve(10, items)

print("M = 15")
solve(15, items)
```
