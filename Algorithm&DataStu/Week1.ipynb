```python
from collections import defaultdict

def AminusB(N,target,nums):
    ans=0
    hash=defaultdict(int)
    for num in nums:
        ans+=hash[num-target]
        hash[num]+=1
    return ans

N,C=map(int,input().split())
nums=list(map(int,input().split()))
nums.sort()
ans=0
P=AminusB(N,C,nums)
print(P)
```
