```python
def ArrayforJosephus(n,k):
    flag=[0]*n
    order=[]
    index=0
    count=0
    num=0
    while count<n:
        if flag[index]==0:
            num+=1
            if num==k:
                order.append(index+1)
                flag[index]=1
                count+=1
                num=0
        index=(index+1)%n
    return order


n,k=map(int ,input().split())
order=ArrayforJosephus(n,k)
print(order[-1])
```

## 递归


```python
def ysf(n, m, i):
    """
    递归求解约瑟夫环：返回第 i 个出列的人的位置（从 0 开始编号）
    :param n: 当前剩余人数
    :param m: 报数间隔
    :param i: 求第几个出列的人（1 ~ n）
    :return: 该人在当前 n 人环中的位置索引（0-based）
    """
    if i == 1:
        # 第一个出列的人：从位置 0 开始数 m-1 步
        return (m - 1 + n) % n
    else:
        # 递归：先求 n-1 人时第 i-1 个出列的位置，再映射回 n 人情况
        return (ysf(n - 1, m, i - 1) + m) % n


def main():
    n, m = map(int, input("请输入总人数 n 和报数间隔 m (空格分隔): ").split())
    
    print("出列顺序（编号从 1 开始）:", end=" ")
    for i in range(1, n + 1):
        # ysf 返回 0-based 索引，+1 转换为 1-based 编号
        position = ysf(n, m, i)
        print(position + 1, end=" ")
    print()  # 换行
    
    # 输出最后剩下的人（即第 n 个出列的人）
    last = ysf(n, m, n) + 1
    print(f"最后剩下的人编号: {last}")


if __name__ == "__main__":
    main()
```

## LinkedList

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
    
class CircleList:
    def __init__(self):
        self.head=None
        self.size=0
    
    def create(self,n):
        if n<=0:
            return
        self.head=Node(1)
        current=self.head
        for i in range(2,n+1):
            new_node=Node(i)
            current.next=new_node
            current=new_node
        current.next=self.head  # 形成循环
        self.size=n

    def Getprev(self):
        current=self.head
        while current.next!=self.head:
            current=current.next
        return current


    def Length(self):
        if not self.head:
            return 0
        count=1
        current=self.head
        while current.next!=self.head:
            count+=1
            current=current.next
        return count

    def josephus(self,k):
        order=[]
        current=self.head
        prev=self.Getprev()

        while self.size>0 :
            for _ in range(k-1):
                prev=current
                current=current.next
            order.append(current.data)
            prev.next=current.next  # 删除当前节点
            current=prev.next  # 移动到下一个节点
            self.size -= 1
        return order
    
n,k=map(int ,input().split())
circle_list=CircleList()
circle_list.create(n)
order=circle_list.josephus(k)
print(order[-1])

```
