# 0-1背包

```python
from typing import List
from functools import cache

def zwknapsack_dfs(capacity: int, w: List[int], v: List[int]) -> int:
    n=len(w)
    @cache
    def dfs(i, c):
        if i<0:
            return 0
        if c<w[i]:
            return dfs(i-1, c)
        return max(dfs(i-1, c), dfs(i-1, c-w[i])+v[i])
    return dfs(n-1, capacity)

ans=zwknapsack_dfs(10, [2, 3, 4, 5], [3, 4, 5, 6])
print(ans)
```

```python
def zwknapsack_dp(capacity: int, w: List[int], v: List[int]) -> int:
    dp=[0]*(capacity+1)
    length=len(w)

    for i in range(length):
        for m in range(capacity,w[i]-1,-1):
            dp[m]=max(dp[m],dp[m-w[i]]+v[i])

    return dp[capacity]
ans=zwknapsack_dp(10, [2, 3, 4, 5], [3, 4, 5, 6])
print(ans)
```

## 恰好装满

```python
from typing import List
from functools import cache
from math import inf

def commknapsack_dfs(capacity: int, w: List[int], v: List[int]) -> int:
    n=len(w)
    @cache
    def dfs(i, c):
        if i<0:
            return 0 if c==0 else -inf
        if c<w[i]:
            return dfs(i-1, c)
        return max(dfs(i-1, c), dfs(i, c-w[i])+v[i])
    return dfs(n-1, capacity)

ans=commknapsack_dfs(10, [2, 3, 4, 5], [2, 4, 5, 6])
print(ans)
```

```python
#完全背包、dp版本
def comknapsack_dp(capacity: int, w: List[int], v: List[int]) -> int:
    n=len(w)
    dp=[0]*(capacity+1)

    for i in range(n):
        for m in range(w[i],capacity+1):
            dp[m]=max(dp[m],dp[m-w[i]]+v[i])
    return dp[capacity]

ans=comknapsack_dp(10, [2, 3, 4, 5], [3, 4, 5, 6])
print(ans)
```

```python
#完全背包、dp版本、恰好
def commknapsack_dp(capacity: int, w: List[int], v: List[int]) -> int:
    n=len(w)
    dp=[-inf]*(capacity+1)
    dp[0]=0


    for i in range(n):
        for m in range(w[i],capacity+1):
            dp[m]=max(dp[m],dp[m-w[i]]+v[i])
    return dp[capacity] if dp[capacity]!=-inf else -1

ans=commknapsack_dp(9, [2, 3, 4, 5], [3, 4, 5, 6])
print(ans)
```

# 最小价值、至少装多少
题目：至少装满背包（最小价值版）\
题意\
有一个背包，要求总重量至少达到 8。有 4 种物品，每种可以无限次选（完全背包）：\
重量：w = [2, 3, 4, 5]\
价值：v = [3, 4, 5, 6]\
求：满足重量 ≥8 的前提下，最小价值是多少？

```python
def min_value_at_least(capacity, weight, value):
    INF = float('inf')
    # dp[j] 表示：恰好装到重量 j 的最小价值
    dp = [INF] * (capacity + 1)
    dp[0] = 0  # 重量0，价值0

    for w, v in zip(weight, value):
        # 完全背包：正序遍历
        for j in range(capacity + 1):
            if dp[j] == INF:
                continue  # 这个状态不可达，跳过

            # 选这个物品后，新重量
            new_j = j + w

            if new_j <= capacity:
                # 没超过容量，更新对应位置
                dp[new_j] = min(dp[new_j], dp[j] + v)
            else:
                # 超过了，也算作达到“至少capacity”，更新 dp[capacity]
                dp[capacity] = min(dp[capacity], dp[j] + v)

    return dp[capacity]


# 测试
weight = [2, 3, 4, 5]
value = [3, 4, 5, 6]
print(min_value_at_least(8, weight, value))
```

```python
def knapsack_01_at_least(capacity, weight, value):
    INF = float('inf')
    # dp[j] = 恰好装到重量 j 的最小价值
    dp = [INF] * (capacity + 1)
    dp[0] = 0

    n = len(weight)
    for i in range(n):
        w = weight[i]
        v = value[i]
        # 01 背包必须倒序遍历，防止重复选
        for j in range(capacity, -1, -1):
            if dp[j] == INF:
                continue
            new_j = j + w
            if new_j <= capacity:
                dp[new_j] = min(dp[new_j], dp[j] + v)
            else:
                # 超过容量也算满足“至少”，统一记在 dp[capacity]
                dp[capacity] = min(dp[capacity], dp[j] + v)

    return dp[capacity]

# 测试
w = [2, 3, 4, 5]
v = [3, 4, 5, 6]
print(knapsack_01_at_least(8, w, v))
```

```python
def zero1_at_least(capacity, weight, value):
    INF = float('inf')
    dp = [INF] * (capacity + 1)
    dp[0] = 0

    for w, v in zip(weight, value):
        # 01 背包 → 倒序遍历
        for j in range(capacity, -1, -1):
            if dp[j] == INF:
                continue
            
            # ✅ 这里就是你要的：一行代替 if else！
            dp[min(j + w, capacity)] = min(dp[min(j + w, capacity)], dp[j] + v)

    return dp[capacity]

# 测试
weight = [2,3,4,5]
value  = [3,4,5,6]
print(zero1_at_least(8, weight, value))  # 输出 10
```
