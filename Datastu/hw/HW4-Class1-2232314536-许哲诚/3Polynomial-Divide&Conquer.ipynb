```python
import random
import time
THRESHOLD = 32

def naive_multiply(A, B):
    n = len(A)
    m = len(B)
    result = [0] * (n + m - 1)
    
    for i in range(n):
        if A[i] == 0: continue
        for j in range(m):
            result[i + j] += A[i] * B[j]
    return result

def add_poly(A, B):
    result = [0] * max(len(A), len(B))
    for i in range(len(A)):
        result[i] += A[i]
    for i in range(len(B)):
        result[i] += B[i]
    return result

def sub_poly(A, B):
    result = [0] * max(len(A), len(B))
    for i in range(len(A)):
        result[i] += A[i]
    for i in range(len(B)):
        result[i] -= B[i]
    return result

def karatsuba(A, B):
    n = max(len(A), len(B))
    if n <= THRESHOLD:
        return naive_multiply(A, B)
    
    if n % 2 != 0:
        n += 1
    A = A + [0] * (n - len(A))
    B = B + [0] * (n - len(B))
    
    m = n // 2
    
    A_low = A[:m]
    A_high = A[m:]
    
    B_low = B[:m]
    B_high = B[m:]
    
    z0 = karatsuba(A_low, B_low)
    z2 = karatsuba(A_high, B_high)
    
    sum_A = add_poly(A_low, A_high)
    sum_B = add_poly(B_low, B_high)
    z1_temp = karatsuba(sum_A, sum_B)
    
    z1 = sub_poly(z1_temp, z0)
    z1 = sub_poly(z1, z2)
    
    result = [0] * (2 * n - 1)
    
    for i in range(len(z0)):
        result[i] += z0[i]
    for i in range(len(z1)):
        result[i + m] += z1[i]
    for i in range(len(z2)):
        result[i + 2 * m] += z2[i]
        
    return result

def generate_polynomial(degree):
    return [random.randint(-20, 20) for _ in range(degree)]

def poly_to_string(coeffs):
    if not coeffs:
        return "0"
    terms = []
    for i, c in enumerate(coeffs):
        if c == 0:
            continue
        
        abs_c = abs(c)
        sign = "-" if c < 0 else "+"

        term_str = ""
        if i == 0:
            term_str = f"{abs_c}"
        elif i == 1:
            if abs_c == 1:
                term_str = "x"
            else:
                term_str = f"{abs_c}x"
        else:
            if abs_c == 1:
                term_str = f"x^{i}"
            else:
                term_str = f"{abs_c}x^{i}"

        if i == 0:
            terms.append(f"{-abs_c}" if c < 0 else f"{abs_c}")
        else:
            terms.append(f" - {term_str}" if c < 0 else f" + {term_str}")
    if not terms:
        return "0"
    return "".join(terms)

def main():
    try:
        n_input = input("\n请输入多项式的长度 n (建议 10-200 以便观察表达式): ")
        n = int(n_input)
        if n < 1:
            print("n 必须大于 0，默认设置为 10")
            n = 10
    except ValueError:
        print("输入无效，默认设置为 10")
        n = 10

    print(f"\n正在生成 {n} 项随机多项式...")
    
    A = generate_polynomial(n)
    B = generate_polynomial(n)
    C = generate_polynomial(n)
    

    while len(A) > 1 and A[-1] == 0: A.pop()
    while len(B) > 1 and B[-1] == 0: B.pop()
    while len(C) > 1 and C[-1] == 0: C.pop()
    
    print("\n--- Input Polynomials ---")
    print(f"A(x) = {poly_to_string(A)}")
    print(f"B(x) = {poly_to_string(B)}")
    print(f"C(x) = {poly_to_string(C)}")

    print("\n--- Calculation & Performance ---")
    
    start_time = time.time()
    Temp = karatsuba(A, B)
    D = karatsuba(Temp, C)
    end_time = time.time()
    
    elapsed_ms = (end_time - start_time) * 1000
    print(f"Computation Time: {elapsed_ms:.4f} ms")
    print(f"Result Degree: {len(D)-1}")
    
    print("\n--- Result D(x) = A(x)·B(x)·C(x) ---")
    
    if len(D) <= 50:
        print(f"D(x) = {poly_to_string(D)}")
    else:
        print("D(x) expression is too long to display fully.")
        print(f"First 5 terms: {poly_to_string(D[:5])} ...")
        print(f"Last 5 terms: ... + {poly_to_string(D[-5:])}")
        print("(Full coefficient list saved in variable D)")

if __name__ == "__main__":
    main()
```

```python
def performance_comparison(sizes):
    """
    对比朴素算法和Karatsuba算法在不同规模下的性能
    sizes: 要测试的多项式长度列表
    """
    print("\n" + "="*70)
    print("Performance Comparison: Naive vs Karatsuba")
    print("="*70)
    print(f"{'Size':<8} {'Naive(ms)':<12} {'Karatsuba(ms)':<15} {'Speedup':<10} {'Status'}")
    print("-"*70)
    
    results = []
    
    for n in sizes:
        # 生成随机测试数据
        A = generate_polynomial(n)
        B = generate_polynomial(n)
        
        # 测试 Karatsuba 算法 (始终执行)
        start = time.time()
        _ = karatsuba(A.copy(), B.copy())
        kara_time = (time.time() - start) * 1000
        
        # 测试朴素算法 (仅当 n <= 1024 时执行，避免等待过久)
        if n <= 1024:
            start = time.time()
            _ = naive_multiply(A.copy(), B.copy())
            naive_time = (time.time() - start) * 1000
            speedup = naive_time / kara_time if kara_time > 0 else float('inf')
            status = "✓"
        else:
            naive_time = None
            speedup = None
            status = "Skipped (too slow)"
        
        # 格式化输出
        if naive_time is not None:
            print(f"{n:<8} {naive_time:<12.4f} {kara_time:<15.4f} {speedup:<10.2f}x {status}")
        else:
            print(f"{n:<8} {'N/A':<12} {kara_time:<15.4f} {'N/A':<10} {status}")
        
        results.append({
            'size': n,
            'naive': naive_time,
            'karatsuba': kara_time,
            'speedup': speedup
        })
    
    print("-"*70)
    return results
```

```python
run_compare = input("\n是否运行性能对比测试？(y/n, 默认n): ").strip().lower()
    
if run_compare == 'y':
        # 执行性能对比：64, 128, 512, 1024, 4096
    test_sizes = [64, 128, 512, 1024, 4096]
    performance_comparison(test_sizes)
```
