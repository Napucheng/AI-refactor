## 最大子数组和

### DP

```python
import random
def max_subarray_sum(arr):
    max_sum = float('-inf')
    current_sum = 0
    start=0
    temporary_start=0
    end=0
    length=len(arr)

    for i in range(len(arr)):
        current_sum += arr[i]

        if current_sum > max_sum:
            max_sum = current_sum
            start = temporary_start
            end = i

        if current_sum < 0:
            current_sum = 0
            temporary_start = i + 1
    return max_sum, start, end
# Example usage
arr = [random.randint(-100, 100) for _ in range(200)]
print(f"Input array: {arr}")
result = max_subarray_sum(arr)
print(f"Maximum subarray sum: {result[0]}, Start index: {result[1]}, End index: {result[2]}")
```

### 分治

```python
def max_subarray_sum(arr):
    def divide(left,right):
        if left == right:
            return arr[left]

        mid = (left + right) // 2

        left_sum = divide(left, mid)
        right_sum = divide(mid + 1, right)

        cross_sum = max_crossing_sum(left, mid, right)

        return max(left_sum, right_sum, cross_sum)

    def max_crossing_sum(left, mid, right):
        left_sum = float('-inf')
        current_sum = 0
        for i in range(mid, left - 1, -1):
            current_sum += arr[i]
            left_sum = max(left_sum, current_sum)

        right_sum = float('-inf')
        current_sum = 0
        for i in range(mid + 1, right + 1):
            current_sum += arr[i]
            right_sum = max(right_sum, current_sum)

        return left_sum + right_sum

    if not arr:
        return 0

    return divide(0, len(arr) - 1)

# Example usage
if __name__ == "__main__":
    arr = [-2,90,-3,4,-1,2,1,-5,4]
    print("Maximum subarray sum is", max_subarray_sum(arr))
```

### 扩展到二维？

```python
matrix=[[1, 2, 3],
        [4, -5, 6],
        [7, 8, -9]]
def max_submatrix_sum(matrix):
    if not matrix or not matrix[0]:
        return 0

    max_sum = float('-inf')
    rows = len(matrix)
    cols = len(matrix[0])

    for left in range(cols):
        temp = [0] * rows
        for right in range(left, cols):
            for i in range(rows):
                temp[i] += matrix[i][right]

            current_max_sum = max_subarray_sum(temp)
            max_sum = max(max_sum, current_max_sum)

    return max_sum
# Example usage
print("Maximum submatrix sum is", max_submatrix_sum(matrix))
```
