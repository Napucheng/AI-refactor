## 栈计算表达式

```python
def evaluate(expression):
    nums = []      # 操作数栈
    ops = []       # 运算符栈

    def apply_op():
        b = nums.pop()
        a = nums.pop()
        op = ops.pop()

        if op == '+':
            nums.append(a + b)
        elif op == '-':
            nums.append(a - b)
        elif op == '*':
            nums.append(a * b)
        elif op == '/':
            nums.append(a / b)

    def priority(op):
        if op in '+-':
            return 1
        if op in '*/':
            return 2
        return 0

    i = 0
    while i < len(expression):

        if expression[i] == ' ':
            i += 1
            continue

        # 读取数字
        if expression[i].isdigit():
            num = 0
            while i < len(expression) and expression[i].isdigit():
                num = num * 10 + int(expression[i])
                i += 1
            nums.append(num)
            continue

        # 左括号
        if expression[i] == '(':
            ops.append(expression[i])

        # 右括号
        elif expression[i] == ')':
            while ops and ops[-1] != '(':
                apply_op()
            ops.pop()

        # 运算符
        else:
            while ops and priority(ops[-1]) >= priority(expression[i]):
                apply_op()
            ops.append(expression[i])

        i += 1

    # 处理剩余运算
    while ops:
        apply_op()

    return nums[0]


expr = input("请输入表达式: ")
print("结果:", evaluate(expr))
```

## 逆波兰表达式
可以仅使用一个栈完成计算，如下是普通表达式转换逆波兰表达式的封装函数和一个栈完成逆波兰表达式计算的函数。

```python
def eval_rpn(tokens):
    stack = []

    for token in tokens:
        if token in "+-*/":
            b = stack.pop()
            a = stack.pop()

            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                stack.append(a / b)
        else:
            stack.append(int(token))

    return stack[0]


expr = input()
tokens = list(expr)   # 逐字符解析

print(eval_rpn(tokens))
```

```python
def infix_to_rpn(expression):
    stack = []
    output = []

    priority = {'+':1, '-':1, '*':2, '/':2}

    for token in expression.split():

        if token.isdigit():
            output.append(token)

        elif token in "+-*/":
            while stack and stack[-1] != '(' and priority[stack[-1]] >= priority[token]:
                output.append(stack.pop())
            stack.append(token)

        elif token == '(':
            stack.append(token)

        elif token == ')':
            while stack[-1] != '(':
                output.append(stack.pop())
            stack.pop()

    while stack:
        output.append(stack.pop())

    return output
```
