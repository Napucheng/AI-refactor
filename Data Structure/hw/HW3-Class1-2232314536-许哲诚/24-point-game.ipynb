# 24点游戏求解器

使用逆波兰表达式和栈来判断四个数能否通过加减乘除和括号组成24，并输出所有可能的中缀表达式。

```python
import itertools
from collections import deque
```

```python
def evaluate_rpn(tokens):
    stack = []
    for token in tokens:
        if token in '+-*/':
            if len(stack) < 2:
                return None
            b = stack.pop()
            a = stack.pop()
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                if b == 0:
                    return None
                stack.append(a / b)
        else:
            try:
                stack.append(float(token))
            except ValueError:
                return None
    return stack[0] if len(stack) == 1 else None
```

```python
def generate_rpn_expressions(nums):
    ops = ['+', '-', '*', '/']
    expressions = []
    
    # 数字全排列
    for num_perm in itertools.permutations(nums):
        a, b, c, d = num_perm
        
        # 运算符组合
        for op1, op2, op3 in itertools.product(ops, repeat=3):
            # 5种括号结构对应的后缀表达式
            # 1. ((a op1 b) op2 c) op3 d -> a b op1 c op2 d op3
            expressions.append([str(a), str(b), op1, str(c), op2, str(d), op3])
            
            # 2. (a op1 (b op2 c)) op3 d -> a b c op2 op1 d op3
            expressions.append([str(a), str(b), str(c), op2, op1, str(d), op3])
            
            # 3. a op1 ((b op2 c) op3 d) -> a b c op2 d op3 op1
            expressions.append([str(a), str(b), str(c), op2, str(d), op3, op1])
            
            # 4. a op1 (b op2 (c op3 d)) -> a b c d op3 op2 op1
            expressions.append([str(a), str(b), str(c), str(d), op3, op2, op1])
            
            # 5. (a op1 b) op2 (c op3 d) -> a b op1 c d op3 op2
            expressions.append([str(a), str(b), op1, str(c), str(d), op3, op2])
    
    return expressions
```

```python
def find_24_expressions(nums):
    expressions = generate_rpn_expressions(nums)
    valid_expressions = []
    for rpn in expressions:
        result = evaluate_rpn(rpn)
        if result is not None and abs(result - 24) < 1e-6:  # 浮点精度
            valid_expressions.append(rpn)
    return valid_expressions
```

```python
def rpn_to_infix(rpn):
    stack = []
    for token in rpn:
        if token in '+-*/':
            b = stack.pop()
            a = stack.pop()
            stack.append(f"({a}{token}{b})")
        else:
            stack.append(token)
    return stack[0]
```

```python
# 示例测试
nums = [9, 2, 10, 4]
valid_rpn = find_24_expressions(nums)
if valid_rpn:
    print("可以组成24的表达式：")
    for rpn in valid_rpn:
        infix = rpn_to_infix(rpn)
        print(f"{infix}=24")
else:
    print("无法组成24")
```
